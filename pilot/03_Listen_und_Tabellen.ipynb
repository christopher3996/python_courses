## 4 · Bonus: Dictionaries

Ein **Dictionary** ordnet Schlüsseln Werte zu. Schlüssel sind eindeutig. Einträge behalten in Python ihre Einfügereihenfolge; ein Dictionary ist aber nicht automatisch sortiert.

```python
student = {"name": "Max Mustermann", "matrikelnummer": "12345678"}
print(student["name"])
student["fachbereich"] = "Wirtschaft"
student["name"] = "Alice"
del student["fachbereich"]
if "name" in student:
    print("Name vorhanden")
```

Die eckigen Klammern enthalten hier einen Schlüssel statt eines Listenindex.
